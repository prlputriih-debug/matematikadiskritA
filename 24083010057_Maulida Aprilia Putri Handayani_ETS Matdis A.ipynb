# 2. Heatmap Frekuensi dan Cacah Kata

```python
import seaborn as sns
import matplotlib.pyplot as plt

#Hitung frekuensi kombinasi konsonan-vokal
frekuensi = df_artikel.groupby(['Konsonan', 'Vokal']).size().reset_index(name='Frekuensi')

# Ubah menjadi pivot table untuk heatmap
pivot_freq = frekuensi.pivot(index='Konsonan', columns='Vokal', values='Frekuensi').fillna(0)

#Hitung jumlah kata unik dalam tiap kombinasi konsonan-vokal
pivot_count = df_artikel.groupby(['Konsonan', 'Vokal'])['Kata'].nunique().unstack(fill_value=0)

#Visualisasi Heatmap

plt.figure(figsize=(12, 5))

# Heatmap 1: Frekuensi kombinasi konsonan-vokal
plt.subplot(1, 2, 1)
sns.heatmap(pivot_freq, annot=True, fmt=".0f", cmap="YlGnBu")
plt.title("Frekuensi Kombinasi Konsonan-Vokal")
plt.xlabel("Jumlah Vokal")
plt.ylabel("Jumlah Konsonan")

# Heatmap 2: Jumlah kata per kombinasi konsonan-vokal
plt.subplot(1, 2, 2)
sns.heatmap(pivot_count, annot=True, fmt=".0f", cmap="OrRd")
plt.title("Cacah Kata per Kombinasi Konsonan-Vokal")
plt.xlabel("Jumlah Vokal")
plt.ylabel("Jumlah Konsonan")

plt.tight_layout()
plt.show()
```

Dari kedua heatmap tersebut, dapat dilihat bahwa heatmap pertama merupakan frekuensi komnbinasi konsonan vokal, pada heatmap pertama warna paling gelap adalah biru tua dengan nilai 84 yang artinya frekuensi pada kombinasi jumlah konsonan 3 dan vokal 2 paling banyak disusul dengan nilai 64 pada konsonan 2 dan vokal 2, sedangkan nilai terendah yaitu berwarna semakin terang ada pada jumlah konsonan 7 dan vokal 3, konsonan 9 dan vokal 4 serta konsonan 9 vokal 6 yang masing-masing memiliki 1 frekuensi kemunculan.

Pada heatmap kedua yang merupakan cacah kata yang masuk dalam kombinasi konsonan vokal didapatkan hasil paling gelap berwarna merah pada konsonan 3 dan vokal 2, yang artinya pada kombinasi ini cukup beragam kata yang ada, dengan nilai sebesar 35 sedangkan nilai terkecil ada pada beberapa kombinasi lain contohnya seperti kombinasi konsonan 1 dan vokal 1.

```python
#Grupkan berdasarkan kombinasi konsonan–vokal
grup_kombinasi = df_artikel.groupby(['Konsonan', 'Vokal'])['Kata'].apply(list).reset_index(name='Daftar_Kata')
grup_kombinasi['Frekuensi Kata'] = grup_kombinasi['Daftar_Kata'].apply(len)

#Tampilkan tabel lengkap
print("\n=== DAFTAR KATA PER KOMBINASI KONSONAN–VOKAL ===")
print(grup_kombinasi)
```

```python
from statistics import mean

paragraf_list = [p.strip() for p in teks.split('\n') if p.strip()]

data = []

for i, paragraf in enumerate(paragraf_list, start=1):
    kalimat_list = [k.strip() for k in re.split(r'[.!?]', paragraf) if k.strip()] #Pisahkan kalimat dengan tanda titik, seru, atau tanya
    jumlah_kalimat = len(kalimat_list)

    jumlah_kata_per_kalimat = [len(k.split()) for k in kalimat_list] # Hitung rata-rata jumlah kata per kalimat
    rata_kata_per_kalimat = mean(jumlah_kata_per_kalimat) if jumlah_kalimat > 0 else 0

    total_kata = sum(jumlah_kata_per_kalimat)

    data.append({
        "Paragraf_ke": i,
        "Jumlah_Kalimat": jumlah_kalimat,
        "Rata-rata_Kata_per_Kalimat": round(rata_kata_per_kalimat, 2),
        "Jumlah_Kata_Paragraf": total_kata
    })

df = pd.DataFrame(data)
print("=== HASIL ANALISIS STRUKTUR TEKS ===")
print(df)

plt.figure(figsize=(8, 5))
plt.plot(df['Paragraf_ke'], df['Rata-rata_Kata_per_Kalimat'], marker='o', color='orange')
plt.title('Rata-rata Kata per Kalimat di Tiap Paragraf')
plt.xlabel('Paragraf ke-')
plt.ylabel('Rata-rata Jumlah Kata per Kalimat')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
from collections import Counter

#Bersihkan teks dari tanda baca & ubah ke huruf kecil
kata = re.findall(r'\b[a-zA-Záéíóúüñçàèìòùäëïöü]+\b', teks.lower())

#Hitung frekuensi kata
frekuensi = Counter(kata)

#Ambil 15 kata paling sering muncul
top_words = frekuensi.most_common(15)

#Visualisasi frekuensi kata
plt.figure(figsize=(10,5))
plt.barh([w[0] for w in top_words][::-1], [w[1] for w in top_words][::-1], color='lightgreen')
plt.xlabel("Frekuensi")
plt.ylabel("Kata")
plt.title("15 Kata Paling Sering Muncul")
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()

def hitung_vokal_konsonan(kata):
    vokal = sum(1 for huruf in kata if huruf in "aeiou")
    konsonan = sum(1 for huruf in kata if huruf.isalpha() and huruf not in "aeiou")
    return vokal, konsonan

data = []
for k in set(kata):
    v, c = hitung_vokal_konsonan(k)
    data.append((k, v, c))

df = pd.DataFrame(data, columns=["Kata", "Vokal", "Konsonan"])

plt.figure(figsize=(8,5))
plt.scatter(df["Konsonan"], df["Vokal"], alpha=0.6, color='coral')
plt.xlabel("Jumlah Konsonan")
plt.ylabel("Jumlah Vokal")
plt.title("Sebaran Rasio Vokal dan Konsonan Tiap Kata")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

print("=== Analisis Pola Bahasa ===")
print(f"Total kata unik: {len(df)}")
print(f"Kata paling sering: {top_words[0][0]} (muncul {top_words[0][1]} kali)")
rata_vokal = df['Vokal'].mean()
rata_konsonan = df['Konsonan'].mean()
print(f"Rata-rata huruf vokal per kata: {rata_vokal:.2f}")
print(f"Rata-rata huruf konsonan per kata: {rata_konsonan:.2f}")
```

Selain itu rata-rata huruf vokal yang muncul sebanyak 3,05 huruf dan konsonan sebanyak 4,14 huruf. 

Grafik scatter plot menunjukkan penyebaran kata berdasarkan jumlah konsonan dan vokal, pada titik-titik yang menyebar di area menengah menunjukkan bahwa sebagian besar kata memiliki panjang sedang dan struktur seimbang.

```python
import textstat

flesch_score = textstat.flesch_reading_ease(teks)
grade_level = textstat.flesch_kincaid_grade(teks)
total_kalimat = textstat.sentence_count(teks)
total_kata = textstat.lexicon_count(teks)
rata_panjang_kalimat = total_kata / total_kalimat

print("=== Analisis Keterbacaan Teks ===")
print(f"Indeks Flesch Reading Ease : {flesch_score:.2f}")
print(f"Tingkat pendidikan yang sesuai (Flesch-Kincaid Grade): {grade_level:.2f}")
print(f"Jumlah kalimat: {total_kalimat}")
print(f"Jumlah kata: {total_kata}")
print(f"Rata-rata panjang kalimat: {rata_panjang_kalimat:.2f} kata per kalimat")

if flesch_score >= 80:
    interpretasi = "Teks sangat mudah dibaca (cocok untuk siswa SD)."
elif flesch_score >= 60:
    interpretasi = "Teks cukup mudah dibaca (cocok untuk pelajar SMP)."
elif flesch_score >= 40:
    interpretasi = "Teks agak sulit dibaca (cocok untuk pelajar SMA)."
else:
    interpretasi = "Teks sulit dibaca (cocok untuk mahasiswa atau akademisi)."

print("\nInterpretasi:", interpretasi)
```
